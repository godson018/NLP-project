## NLP Project 7: Hotel Review Analysis


### Problem Statement

A hotel management company wants to analyze customer reviews using NLP techniques to understand customer satisfaction and identify service-related issues.

Tasks
1. Collect hotel review data.
2. Preprocess guest reviews.
3. Perform sentiment analysis.
4. Identify service-related issues.
5. Generate customer satisfaction report.

Step 1: Import Required Libraries

```python
import pandas as pd
import nltk
import string

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from textblob import TextBlob
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

Step 2: Create the Hotel Review Dataset

```python
data = {
    "Review": [
        "The room was clean and spacious. Staff were very friendly.",
        "Food was delicious and service was excellent.",
        "The room was dirty and bathroom smelled bad.",
        "Staff were rude and check-in took too long.",
        "Hotel location is good but WiFi is slow.",
        "Breakfast was average and rooms were comfortable.",
        "AC was not working properly.",
        "Swimming pool was clean and well maintained."
    ]
}

df = pd.DataFrame(data)

df
```

Step 3: Preprocess the Reviews

```python
lemmatizer = WordNetLemmatizer()

def preprocess(text):

    # Convert to lowercase
    text = text.lower()

    # Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))

    # Tokenize
    words = word_tokenize(text)

    # Remove stop words
    words = [word for word in words if word not in stopwords.words('english')]

    # Lemmatization
    words = [lemmatizer.lemmatize(word) for word in words]

    return " ".join(words)
```

Step 4: Apply Preprocessing

```python
import nltk
nltk.download('punkt_tab')
df["Processed Review"] = df["Review"].apply(preprocess)

df
```

Step 5: Perform Sentiment Analysis

```python
def get_sentiment(text):

    score = TextBlob(text).sentiment.polarity

    if score > 0:
        return "Positive"

    elif score < 0:
        return "Negative"

    else:
        return "Neutral"
```

The `get_sentiment` function uses `TextBlob` to calculate the polarity of the processed review and assigns 'Positive', 'Negative', or 'Neutral' based on the score.

```python
df["Sentiment"] = df["Processed Review"].apply(get_sentiment)
display(df)
```

```python
df["Sentiment"] = df["Processed Review"].apply(get_sentiment)

df
```

Step 6: Identify Service-Related Issues

```python
issues = {
    "Room": ["room", "bed", "bathroom", "ac"],
    "Staff": ["staff", "service", "reception"],
    "Food": ["food", "breakfast", "restaurant"],
    "Cleanliness": ["clean", "dirty", "smell"],
    "Amenities": ["wifi", "pool", "gym"]
}
```

```python
def identify_issue(review):

    found = []

    for category, keywords in issues.items():

        for word in keywords:

            if word in review:

                found.append(category)

                break

    if len(found) == 0:
        return "No Issue"

    return ", ".join(found)
```

```python
df["Service Issue"] = df["Processed Review"].apply(identify_issue)

df
```

Step 7: Count Sentiments

```python
sentiment_count = df["Sentiment"].value_counts()

print(sentiment_count)
```

Step 8: Count Service Issues

```python
issue_count = df["Service Issue"].value_counts()

print(issue_count)
```

Step 9: Generate Customer Satisfaction Report

```python
print("=========== CUSTOMER SATISFACTION REPORT ===========")

print("\nTotal Reviews:", len(df))

print("\nSentiment Summary")

print(df["Sentiment"].value_counts())

print("\nService Issues")

print(df["Service Issue"].value_counts())
```
